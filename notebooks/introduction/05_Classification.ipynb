Now that cross-validation has indicated the best hyperparameters, let’s build and train the actual model. Let's use the best params returned by the cross-validation:

```py
}
```

````python
import pandas as pd
import numpy as np
import tensorflow as tf
````

````python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, InputLayer, Dropout
````

````python
from ucimlrepo import fetch_ucirepo
from sklearn.preprocessing import StandardScaler

breast_cancer_wisconsin_diagnostic = fetch_ucirepo(id=17) 

X = breast_cancer_wisconsin_diagnostic.data.features
y = breast_cancer_wisconsin_diagnostic.data.targets

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
y = y["Diagnosis"].map({"B": 0, "M": 1}).to_frame()
````

````python
classifier = Sequential()
classifier.add(InputLayer(shape=(30,)))
classifier.add(Dense(units=8, activation='relu', kernel_initializer='random_uniform'))
classifier.add(Dropout(0.2))
classifier.add(Dense(units=8, activation='relu', kernel_initializer='random_uniform'))
classifier.add(Dropout(0.2))
classifier.add(Dense(units=1, activation='sigmoid'))
````

````python
classifier.compile(optimizer='sgd', loss='binary_crossentropy', metrics=['binary_accuracy'])
````

````python
classifier.fit(X_scaled, y, batch_size=10, epochs=50)
````
